## GPT 모델 토큰 대비 성능 비교

아래 5개 모델을 동일한 프롬프트로 실행하고 **토큰 사용량 대비 응답 품질**을 비교합니다.

| 모델 | 특징 |
|------|------|
| `gpt-4o-mini` | GPT-4o 경량 버전 |
| `gpt-4.1-nano` | GPT-4.1 초경량 버전 |
| `gpt-4.1-mini` | GPT-4.1 경량 버전 |
| `gpt-5-nano` | GPT-5 초경량 버전 |
| `gpt-5-mini` | GPT-5 경량 버전 |

```python
# 비교할 모델 목록
MODELS_TO_COMPARE = [
    "gpt-4o-mini",
    "gpt-4.1-nano",
    "gpt-4.1-mini",
    "gpt-5-nano",
    "gpt-5-mini",
]

SYSTEM_PROMPT = """당신은 금융 전문 AI 어시스턴트입니다.
사용자의 소비 패턴을 분석하고 간결하고 실용적인 피드백을 제공합니다."""

USER_PROMPT = """이번 달 소비 현황입니다:
- 식비: 450,000원 (예산 400,000원 → 12.5% 초과)
- 교통비: 85,000원 (예산 100,000원 → 15% 절약)
- 여가비: 230,000원 (예산 150,000원 → 53% 초과)
- 쇼핑: 120,000원 (예산 100,000원 → 20% 초과)
- 저축: 300,000원

위 소비 내역을 분석하고 다음 달을 위한 핵심 조언 3가지를 알려주세요."""

print(f"비교 대상 모델: {MODELS_TO_COMPARE}")
print(f"\n[시스템 프롬프트]\n{SYSTEM_PROMPT}")
print(f"\n[사용자 프롬프트]\n{USER_PROMPT}")
```

```python
import time
from typing import Any
import os

from langsmith import traceable
from openai import OpenAI

openai_client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

@traceable(name="model-token-comparison", tags=["model-comparison", "token-benchmark"])
def run_model_comparison(model, system_prompt, user_prompt):
    """단일 모델에 대해 프롬프트를 실행하고 토큰 사용량 및 응답 결과를 반환한다."""
    start_time = time.perf_counter()
    kwargs = {
        "model": model,
        "messages": [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt},
        ],
    }
    if "gpt-5" in model:
        kwargs["max_completion_tokens"] = 512
        kwargs["temperature"] = 1.0
    else:
        kwargs["max_tokens"] = 512
        kwargs["temperature"] = 0.3
    response = openai_client.chat.completions.create(**kwargs)
    latency = time.perf_counter() - start_time
    usage = response.usage
    p_tok = usage.prompt_tokens if usage else 0
    c_tok = usage.completion_tokens if usage else 0
    t_tok = usage.total_tokens if usage else 0
    spd = round(c_tok / latency, 2) if latency > 0 else 0.0
    return {
        "model": model,
        "prompt_tokens": p_tok,
        "completion_tokens": c_tok,
        "total_tokens": t_tok,
        "latency_sec": round(latency, 3),
        "tokens_per_sec": spd,
        "response": response.choices[0].message.content or "",
    }

print("모델 비교 함수 정의 완료 ✅")
```

```python
results = []
errors = []

for model_name in MODELS_TO_COMPARE:
    print(f"\n🔄 {model_name} 실행 중...")
    try:
        result = run_model_comparison(
            model=model_name,
            system_prompt=SYSTEM_PROMPT,
            user_prompt=USER_PROMPT,
        )
        results.append(result)
        print(
            f"  ✅ 완료 | "
            f"총 토큰: {result['total_tokens']} "
            f"(입력 {result['prompt_tokens']} / 출력 {result['completion_tokens']}) | "
            f"지연: {result['latency_sec']}초 | "
            f"속도: {result['tokens_per_sec']} tok/s"
        )
    except Exception as e:
        error_msg = str(e)
        errors.append({"model": model_name, "error": error_msg})
        print(f"  ❌ 오류: {error_msg}")

print(f"\n실행 완료: {len(results)}개 성공 / {len(errors)}개 실패")
```

## 섹션 2: 실제 프로젝트 파이프라인 기반 모델 성능 비교

임의 프롬프트가 아닌 **실제 Catcher 소비 피드백 파이프라인**으로 5개 모델을 비교합니다.

### 비교 전략
1. **분석 → 해석 → RAG 검색** 단계는 기본 모델로 1회만 실행 (공통 입력 생성)
2. **최종 피드백 체인만** 5개 모델 각각에 실행 (실제 성능 차이 측정)
3. **LangSmith `evaluate()`** 로 자동 품질 점수까지 측정

### 평가 지표
| 지표 | 설명 | 비중 |
|------|------|------|
| `accuracy` | 정확성 — 실제 수치(%) 인용 여부 (데이터 기반 작성 proxy) | 40% |
| `utility` | 구용성 — 구체적 조언/미션/행동 권유 포함 여부 | 30% |
| `format` | 형식 — 금지 패턴 부재 + 적절한 길이 | 30% |
| `llm_quality_score` | GPT-4o-mini가 판단하는 피드백 종합 품질 | (참고) |

```python
import time
from catcher_llm.config.settings import Settings
from catcher_llm.services.consumption_feedback.daily_feedback import (
    generate_daily_feedback,
    make_daily_feedback_input,
)

TEST_CASES = [
    {"member_id": 1, "analysis_date": "2026-04-29", "previous_date": "2026-04-28"},
    {"member_id": 1, "analysis_date": "2026-04-25", "previous_date": "2026-04-24"},
    {"member_id": 1, "analysis_date": "2026-04-20", "previous_date": "2026-04-19"},
]

base_settings = Settings()
shared_inputs = []

print("🔄 공통 파이프라인 입력 생성 중 (분석 → 해석 → RAG)...")
print("   이 단계는 1회만 실행합니다. 약 30~90초 소요됩니다.\n")

for case in TEST_CASES:
    date_label = case["analysis_date"]
    print(f"  📅 {date_label} 처리 중...")
    try:
        baseline = generate_daily_feedback(
            member_id=case["member_id"],
            analysis_date=case["analysis_date"],
            previous_date=case["previous_date"],
            settings=base_settings,
        )
        if baseline.error:
            raise Exception(f"파이프라인 실행 오류: {baseline.error}")
        feedback_input = make_daily_feedback_input(
            user_data=baseline.daily_analysis,
            interpretation_result=baseline.interpretation_result,
            advice_contexts=baseline.retrieved_contexts,
            user_profile=baseline.user_profile,
            memory_context=baseline.memory_context,
        )
        shared_inputs.append({
            "date": date_label,
            "feedback_input": feedback_input,
            "baseline_feedback": baseline.feedback,
        })
        print(f"  ✅ 완료")
    except Exception as e:
        print(f"  ❌ 오류: {e}")

print(f"\n공통 입력 생성 완료: {len(shared_inputs)}개 날짜")
```

```python
import re
from pydantic import SecretStr
from langchain_openai import ChatOpenAI
from catcher_llm.chains.consumption_feedback import build_daily_feedback_chain
from catcher_llm.schemas.consumption_feedback.daily import DailyFeedbackResult

MODELS_TO_COMPARE_PROJ = [
    "gpt-4o-mini",
    "gpt-4.1-nano",
    "gpt-4.1-mini",
    "gpt-5-nano",
    "gpt-5-mini",
]

def _get_temperature(model):
    return 1.0 if "gpt-5" in model else 0.0

def run_feedback_for_model(model, feedback_input, analysis_date):
    """지정 모델로 최종 피드백 체인만 실행하고 결과를 반환한다."""
    llm = ChatOpenAI(
        api_key=SecretStr(os.getenv("OPENAI_API_KEY", "")),
        model=model,
        temperature=_get_temperature(model),
    )
    chain = build_daily_feedback_chain(llm=llm)
    t0 = time.perf_counter()
    result = chain.invoke(feedback_input)
    latency = round(time.perf_counter() - t0, 3)
    if isinstance(result, DailyFeedbackResult):
        return {
            "model": model,
            "analysis_date": analysis_date,
            "latency_sec": latency,
            "scolding_message": result.scolding_message,
            "tomorrow_mission": result.tomorrow_mission,
            "summary_title": result.summary_title,
        }
    raise ValueError(f"Unexpected result type: {type(result)}")

print("피드백 체인 실행 함수 정의 완료 ✅")
```

```python
all_results = []
run_errors = []

print("🔬 모델별 피드백 체인 실행 중...")

for entry in shared_inputs:
    date_label = entry["date"]
    feedback_input = entry["feedback_input"]
    print(f"\n  📅 {date_label}")
    for model in MODELS_TO_COMPARE_PROJ:
        print(f"    {model:<20}", end=" ", flush=True)
        try:
            result = run_feedback_for_model(model, feedback_input, date_label)
            all_results.append(result)
            print(f"✅ {result['latency_sec']:.2f}초 | 피드백 {len(result['scolding_message'])}자")
        except Exception as e:
            run_errors.append({"model": model, "date": date_label, "error": str(e)[:120]})
            print(f"❌ {str(e)[:80]}")

print(f"\n실행 완료: {len(all_results)}개 성공 / {len(run_errors)}개 실패")
```

```python
import pandas as pd
from pydantic import BaseModel, Field
from langchain_openai import ChatOpenAI
import re

# 평가 기준을 위한 LLM 설정 (비용/속도 최적화를 위해 gpt-4o-mini 사용)
eval_llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

class EvalResult(BaseModel):
    accuracy_score: float = Field(..., description="수치의 맥락적 일치성 점수 (0.0 ~ 1.0). 텍스트 내 수치가 데이터 분석 결과와 논리적으로 일치하면 1.0, 부분적으로 맞으면 0.5, 틀리면 0.0.")
    utility_score: float = Field(..., description="조언의 구체성 점수 (0.0 ~ 1.0). '무엇을(대상)', '어떻게(방법)', '얼마나(목표)'가 포함된 완성된 문장이면 1.0, 부족하면 0.0~0.9 부여.")

eval_chain = eval_llm.with_structured_output(EvalResult)

def evaluate_with_llm(feedback_msg: str, user_data_json: str) -> EvalResult:
    prompt = f"""
다음은 사용자의 일일 소비 데이터(JSON)와 이에 대해 생성된 피드백입니다.
아래 두 가지 기준에 따라 평가해주세요.

1. 정확성 (Accuracy): 수치의 맥락적 일치성.
   피드백 텍스트 내에 있는 수치(%)나 금액이 실제 소비 데이터(JSON) 분석 결과와 논리적으로 일치하는가?
2. 유용성 (Utility): 조언의 구체성.
   "무엇을(대상)", "어떻게(방법)", "얼마나(목표)"가 포함된 완성된 문장인가?

[소비 데이터]
{user_data_json}

[피드백 텍스트]
{feedback_msg}
"""
    return eval_chain.invoke(prompt)

# shared_inputs 에서 user_data 추출을 위한 딕셔너리 생성
date_to_data = {item["date"]: item["feedback_input"].get("daily_json", "{}") for item in shared_inputs}

rows = []
print("🤖 LLM을 활용한 정밀 평가 진행 중... (시간이 조금 걸릴 수 있습니다)")
for r in all_results:
    msg = r["scolding_message"]
    analysis_date = r["analysis_date"]
    
    # 해당 날짜의 원본 입력 데이터 가져오기
    user_data_json = date_to_data.get(analysis_date, "{}")
    
    # LLM 기반 평가 실행
    try:
        eval_res = evaluate_with_llm(msg, user_data_json)
        acc_score = eval_res.accuracy_score
        util_score = eval_res.utility_score
    except Exception as e:
        print(f"평가 오류 ({r['model']}): {e}")
        acc_score = 0.0
        util_score = 0.0
        
    rows.append({
        "모델": r["model"], "날짜": analysis_date, "지연(초)": r["latency_sec"],
        "피드백길이": len(msg),
        "정확성": acc_score,
        "구용성": util_score,
        "형식": score_format(msg),
    })

df_detail = pd.DataFrame(rows)
agg = (
    df_detail.groupby("모델")
    .agg(
        평균지연=("지연(초)", "mean"), 평균길이=("피드백길이", "mean"),
        평균정확성=("정확성", "mean"),
        평균구용성=("구용성", "mean"),
        평균형식=("형식", "mean"),
    )
    .round(3).reset_index().sort_values("평균정확성", ascending=False)
)
print("=" * 88)
print("📊 모델별 실제 파이프라인 성능 비교 (3개 날짜 평균, LLM 기반 정밀 평가)")
print("=" * 88)
print(agg.to_string(index=False))
print("=" * 88)
FORBIDDEN_RE = re.compile(r"급증하여|평소\s*\d+[\d.]*%와|diff_point|avg_ratio")
_ACTION_PATS = [
    r"줄여", r"줄이", r"아껴", r"절약", r"목표", r"계획", r"조언",
    r"추천", r"해보", r"미션", r"포인트", r"도전", r"실천", r"관리", r"방법",
]

def score_accuracy(text):
    """정확성 휴리스틱: 수치(%) 포함 여부"""
    return 1.0 if re.search(r"\d+(\.\d+)?%", text) else 0.0

def score_utility(text):
    """구용성 휴리스틱: 행동 유도 키워드 빈도"""
    hits = sum(1 for p in _ACTION_PATS if re.search(p, text))
    return round(min(hits / 3.0, 1.0), 3)

def score_format(text):
    """형식 휴리스틱: 금지 패턴 및 길이"""
    no_forb = 0.0 if FORBIDDEN_RE.search(text) else 1.0
    length_score = min(len(text) / 350.0, 1.0)
    return round(0.6 * no_forb + 0.4 * length_score, 3)
```

```python
from langsmith import Client
from langsmith.evaluation import evaluate as ls_evaluate

ls_client = Client()
EVAL_DATASET_NAME = "catcher-feedback-model-eval"

existing_ds = [d for d in ls_client.list_datasets() if d.name == EVAL_DATASET_NAME]
if existing_ds:
    eval_dataset = existing_ds[0]
    print(f"기존 데이터셋 재사용: {EVAL_DATASET_NAME}")
else:
    eval_dataset = ls_client.create_dataset(
        EVAL_DATASET_NAME,
        description="Catcher 일일 피드백 체인 모델 비교용 데이터셋",
    )
    for entry in shared_inputs:
        ls_client.create_example(
            inputs={"feedback_input": entry["feedback_input"], "analysis_date": entry["date"]},
            outputs={"analysis_date": entry["date"]},
            dataset_id=eval_dataset.id,
        )
    print(f"데이터셋 생성 완료: {EVAL_DATASET_NAME} ({len(shared_inputs)}개 예시)")

judge_llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

def eval_accuracy(run, example):
    """정확성(40%): 실제 수치(%) 인용 여부"""
    msg = (run.outputs or {}).get("scolding_message", "")
    return {"key": "accuracy", "score": 1.0 if re.search(r"\d+(\.\d+)?%", msg) else 0.0}

def eval_utility(run, example):
    """구용성(30%): 구체적 조언/미션/행동어 포함 여부"""
    msg = (run.outputs or {}).get("scolding_message", "")
    _pats = [r"줄여",r"줄이",r"아껴",r"절약",r"목표",r"계획",r"조언",
             r"추천",r"해보",r"미션",r"포인트",r"도전",r"실천",r"관리",r"방법"]
    score = round(min(sum(1 for p in _pats if re.search(p, msg)) / 3.0, 1.0), 3)
    return {"key": "utility", "score": score}

def eval_format(run, example):
    """형식(30%): 금지 패턴 부재(60%) + 적절한 길이(40%)"""
    msg = (run.outputs or {}).get("scolding_message", "")
    no_forb = 0.0 if FORBIDDEN_RE.search(msg) else 1.0
    length_score = min(len(msg) / 350.0, 1.0)
    return {"key": "format", "score": round(0.6 * no_forb + 0.4 * length_score, 3)}

def eval_llm_quality(run, example):
    msg = (run.outputs or {}).get("scolding_message", "")
    if not msg:
        return {"key": "llm_quality_score", "score": 0.0}
    prompt = (
        "아래는 소비 분석 AI의 일일 소비 피드백입니다.\n"
        "다음 기준으로 0.0~1.0 점수를 매겨주세요:\n"
        "1. 구체적인 수치(%)를 근거로 사용했는가 (정확성)\n"
        "2. 내일 실천 가능한 조언이나 미션이 포함되었는가 (구용성)\n"
        "3. 금지 패턴 없이 적절한 길이와 톤인가 (형식)\n\n"
        f"피드백:\n{msg[:600]}\n\n"
        "숫자(0.0~1.0) 하나만 출력하세요."
    )
    try:
        score = float(judge_llm.invoke(prompt).content.strip())
        score = max(0.0, min(1.0, score))
    except Exception:
        score = 0.5
    return {"key": "llm_quality_score", "score": score}

print("평가자 4종 정의 완료 ✅")
print(f"데이터셋: {EVAL_DATASET_NAME} (id={eval_dataset.id})")
```

```python
from typing import Any

eval_experiment_results: dict[str, Any] = {}

for model in MODELS_TO_COMPARE_PROJ:
    print(f"\n🔬 [{model}] LangSmith evaluate 실행 중...")
    _model, _temp = model, _get_temperature(model)

    def _make_target(m, t):
        def _target(inputs):
            _llm = ChatOpenAI(
                api_key=SecretStr(os.getenv("OPENAI_API_KEY", "")),
                model=m, temperature=t,
            )
            _chain = build_daily_feedback_chain(llm=_llm)
            try:
                _r = _chain.invoke(inputs["feedback_input"])
                if isinstance(_r, DailyFeedbackResult):
                    return {"scolding_message": _r.scolding_message,
                            "tomorrow_mission": _r.tomorrow_mission,
                            "summary_title": _r.summary_title}
                return dict(_r) if isinstance(_r, dict) else {"scolding_message": str(_r)}
            except Exception as exc:
                return {"scolding_message": "", "error": str(exc)}
        return _target

    try:
        exp = ls_evaluate(
            _make_target(_model, _temp),
            data=EVAL_DATASET_NAME,
            evaluators=[eval_accuracy, eval_utility, eval_format, eval_llm_quality],
            experiment_prefix=f"feedback-{_model}",
            max_concurrency=1,
        )
        eval_experiment_results[model] = exp
        print(f"  ✅ 완료")
    except Exception as e:
        print(f"  ❌ 오류: {e}")

print("\n모든 모델 LangSmith evaluate 완료 ✅")
print(f"데이터셋: {EVAL_DATASET_NAME} -> Compare 탭")
```

## 섹션 3: 토큰 대비 성능 분석

섹션 2에서 측정한 **품질 점수**와, 이번 섹션에서 새로 캡처하는 **토큰 사용량**을 합쳐  
각 모델의 **토큰 효율(quality per 1K tokens)**을 산출합니다.

### 채점 기준 (종합 품질 산출)
| 영역 | 기준 | 비중 |
|------|------|------|
| 정확성 (Accuracy) | 수치(%) 인용 여부 — 데이터 기반 작성 proxy | 40% |
| 구용성 (Utility) | 구체적 미션/조언 포함 여부 | 30% |
| 형식 (Format) | 금지 패턴 부재 + 적절한 길이 | 30% |

> `shared_inputs`(분석·해석·RAG 결과)는 섹션 2에서 이미 생성된 것을 그대로 재사용합니다.  
> 최종 피드백 체인만 재실행하므로 추가 비용이 최소화됩니다.

```python
import time
from pydantic import SecretStr
from langchain_openai import ChatOpenAI
from langchain_community.callbacks import get_openai_callback
from catcher_llm.chains.consumption_feedback import build_daily_feedback_chain
from catcher_llm.schemas.consumption_feedback.daily import DailyFeedbackResult

# shared_inputs, MODELS_TO_COMPARE_PROJ, _get_temperature -- 섹션 2에서 정의됨

token_records = []

print("🔄 토큰 사용량 캡처를 위해 피드백 체인 재실행 중...")
print("   분석/해석/RAG 단계는 재실행하지 않습니다.\n")

for entry in shared_inputs:
    date_label = entry["date"]
    feedback_input = entry["feedback_input"]
    print(f"  📅 {date_label}")
    for model in MODELS_TO_COMPARE_PROJ:
        print(f"    {model:<20}", end=" ", flush=True)
        llm = ChatOpenAI(
            api_key=SecretStr(os.getenv("OPENAI_API_KEY", "")),
            model=model, temperature=_get_temperature(model),
        )
        chain = build_daily_feedback_chain(llm=llm)
        scolding = ""
        t0 = time.perf_counter()
        try:
            with get_openai_callback() as cb:
                result = chain.invoke(feedback_input)
            latency = round(time.perf_counter() - t0, 3)
            if isinstance(result, DailyFeedbackResult):
                scolding = result.scolding_message
            elif isinstance(result, dict):
                scolding = result.get("scolding_message", "")
            p_tok, c_tok, t_tok = cb.prompt_tokens, cb.completion_tokens, cb.total_tokens
            status = "OK"
        except Exception as exc:
            latency = round(time.perf_counter() - t0, 3)
            p_tok = c_tok = t_tok = 0
            status = f"ERR {str(exc)[:40]}"
        spd = round(c_tok / latency, 1) if latency > 0 and c_tok else 0.0
        print(f"[{status}] 전체 {t_tok:>4}tok (입력 {p_tok}/출력 {c_tok}) | {latency:.2f}초 | {spd} tok/s")
        token_records.append({
            "model": model, "date": date_label,
            "prompt_tokens": p_tok, "completion_tokens": c_tok, "total_tokens": t_tok,
            "latency_sec": latency, "tokens_per_sec": spd, "scolding_message": scolding,
        })

print(f"\n캡처 완료: {len(token_records)}개 기록")
```

```python
import pandas as pd
import re

# 채점 함수 미정의 시 재정의 (섹션 3 단독 실행 대비)
try:
    score_accuracy
except NameError:
    FORBIDDEN_RE = re.compile(r"급증하여|평소\s*\d+[\d.]*%와|diff_point|avg_ratio")
    _ACTION_PATS = [
        r"줄여", r"줄이", r"아껴", r"절약", r"목표", r"계획", r"조언",
        r"추천", r"해보", r"미션", r"포인트", r"도전", r"실천", r"관리", r"방법",
    ]
    def score_accuracy(text):
        return 1.0 if re.search(r"\d+(\.\d+)?%", text) else 0.0
    def score_utility(text):
        hits = sum(1 for p in _ACTION_PATS if re.search(p, text))
        return round(min(hits / 3.0, 1.0), 3)
    def score_format(text):
        no_forb = 0.0 if FORBIDDEN_RE.search(text) else 1.0
        return round(0.6 * no_forb + 0.4 * min(len(text) / 350.0, 1.0), 3)

rows = []
for rec in token_records:
    msg  = rec["scolding_message"]
    acc  = score_accuracy(msg)
    util = score_utility(msg)
    fmt  = score_format(msg)
    # 종합 품질: 정확성 40% + 구용성 30% + 형식 30%
    quality = round(acc * 0.4 + util * 0.3 + fmt * 0.3, 3)
    t_tok   = rec["total_tokens"]
    # 핵심 효율 지표: 1K 토큰당 품질 (높을수록 효율적)
    q_per_1k = round(quality / (t_tok / 1000), 4) if t_tok > 0 else 0.0
    rows.append({
        "모델": rec["model"], "날짜": rec["date"],
        "입력토큰": rec["prompt_tokens"],
        "출력토큰": rec["completion_tokens"],
        "전체토큰": t_tok,
        "지연(초)": rec["latency_sec"],
        "속도(tok/s)": rec["tokens_per_sec"],
        "정확성": acc, "구용성": util, "형식": fmt,
        "종합품질(0-1)": quality,
        "품질/1K토큰": q_per_1k,
    })

df_tp = pd.DataFrame(rows)
agg_tp = (
    df_tp.groupby("모델")
    .agg(
        평균입력토큰=("입력토큰", "mean"),
        평균출력토큰=("출력토큰", "mean"),
        평균전체토큰=("전체토큰", "mean"),
        평균지연=("지연(초)", "mean"),
        평균속도=("속도(tok/s)", "mean"),
        평균정확성=("정확성", "mean"),
        평균구용성=("구용성", "mean"),
        평균형식=("형식", "mean"),
        평균종합품질=("종합품질(0-1)", "mean"),
        품질_1K토큰=("품질/1K토큰", "mean"),
    )
    .round(3).reset_index()
    .sort_values("품질_1K토큰", ascending=False)
    .reset_index(drop=True)
)

print("=" * 105)
print("📊 GPT 모델 토큰 대비 성능 종합표 (정렬: 품질/1K토큰 내림차순)")
print("=" * 105)
print(agg_tp.to_string(index=False))
print("=" * 105)
```

```python
# 랜킹 출력
print("\n== 토큰 효율 랜킹 (품질/1K토큰 기준) ==")
max_q1k = agg_tp["품질_1K토큰"].max()
for rank, (_, row) in enumerate(agg_tp.iterrows(), start=1):
    bar_len = max(1, int(row["품질_1K토큰"] / max_q1k * 20)) if max_q1k > 0 else 1
    bar = "#" * bar_len + "-" * (20 - bar_len)
    print(f"  {rank}. {row['모델']:<18} [{bar}] {row['품질_1K토큰']:.4f}")

print("\n== 토큰 소비량 랜킹 (적을수록 경제적) ==")
max_tok = agg_tp["평균전체토큰"].max()
for rank, (_, row) in enumerate(agg_tp.sort_values("평균전체토큰").iterrows(), start=1):
    bar_len = max(1, int(row["평균전체토큰"] / max_tok * 20)) if max_tok > 0 else 1
    bar = "#" * bar_len + "-" * (20 - bar_len)
    print(f"  {rank}. {row['모델']:<18} [{bar}] {row['평균전체토큰']:.0f} 토큰")

best_eff  = agg_tp.iloc[0]
best_qual = agg_tp.loc[agg_tp["평균종합품질"].idxmax()]
least_tok = agg_tp.loc[agg_tp["평균전체토큰"].idxmin()]
fastest   = agg_tp.loc[agg_tp["평균속도"].idxmax()]

sep = "=" * 72
print(f"\n{sep}")
print("최종 결론")
print(sep)
print(f"  토큰 효율 1위 : {best_eff['모델']:<18} (품질/1K토큰 = {best_eff['품질_1K토큰']:.4f})")
print(f"  절대 품질 1위 : {best_qual['모델']:<18} (종합품질 = {best_qual['평균종합품질']:.3f})")
print(f"  최소 토큰 모델: {least_tok['모델']:<18} (평균 {least_tok['평균전체토큰']:.0f} 토큰)")
print(f"  최고 속도 모델: {fastest['모델']:<18} ({fastest['평균속도']:.1f} tok/s)")
print(sep)
print("\n  품질/1K토큰이 높은 모델 = 같은 비용으로 더 좋은 피드백을 생성하는 모델")
print(f"  LangSmith 상세 실험: 데이터셋 'catcher-feedback-model-eval' -> Compare 탭")
```
