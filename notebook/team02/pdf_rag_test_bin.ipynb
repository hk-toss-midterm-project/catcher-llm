```python
import os
from pathlib import Path
import sys


# 프로젝트 루트 경로를 찾아 src 폴더를 sys.path에 추가합니다.
def find_project_root() -> Path:
    curr = Path.cwd()
    for parent in [curr] + list(curr.parents):
        if (parent / "pyproject.toml").exists():
            return parent
    return curr

PROJECT_ROOT = find_project_root()
src_dir = str(PROJECT_ROOT / "src")
if src_dir not in sys.path:
    sys.path.insert(0, src_dir)

os.environ.setdefault("LANGSMITH_TRACING", "false")
```

```python
from langchain_community.document_loaders import PyPDFLoader

# 프로젝트 루트를 기준으로 경로를 설정합니다.
file_path = str(PROJECT_ROOT / "data/raw/pdf/saving_tips/2.pdf")


loader = PyPDFLoader(file_path=file_path)
docs = loader.load()

print(f"불러온 문서 수: {len(docs)}")
print(docs[1].page_content)
```

```python
print(docs[2].page_content)
print("=" * 100)
print(docs[3].page_content)
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=1,
    chunk_overlap=1,
    length_function=len
)


# 여기를 split_documents로 바꿨어요!
split_docs = text_splitter.split_documents(docs)
split_docs[0]
print(f' ==> [Line 12]: \033[38;2;222;240;227m[split_docs[0]]\033[0m({type(split_docs[0]).__name__}) = \033[38;2;241;199;102m{split_docs[0]}\033[0m')
split_docs[1]
print(f' ==> [Line 14]: \033[38;2;242;81;151m[split_docs[1]]\033[0m({type(split_docs[1]).__name__}) = \033[38;2;179;50;243m{split_docs[1]}\033[0m')
split_docs[2]
print(f' ==> [Line 16]: \033[38;2;57;248;80m[split_docs[2]]\033[0m({type(split_docs[2]).__name__}) = \033[38;2;241;188;84m{split_docs[2]}\033[0m')

print(f"분할된 문서 수: {len(split_docs)}")
```

```python
# 임베딩 

from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
embeddings 
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(docs, embeddings)

vectorstore.index.ntotal
```

```python
retriever = vectorstore.as_retriever()
```

```python
query = "사회초년생을 위한 금융꿀팁 7가지"

# invoke() 메서드로 문서 검색
res = retriever.invoke(query)

len(res)
for i in res:
    print(i.page_content)
    print("=" * 100)
```
