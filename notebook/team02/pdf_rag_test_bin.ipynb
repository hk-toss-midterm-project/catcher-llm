```python
import os
import sys
from pathlib import Path


# 프로젝트 루트 경로를 찾아 src 폴더를 sys.path에 추가합니다.
def find_project_root() -> Path:
    curr = Path.cwd()
    for parent in [curr] + list(curr.parents):
        if (parent / "pyproject.toml").exists():
            return parent
    return curr


PROJECT_ROOT = find_project_root()
src_dir = str(PROJECT_ROOT / "src")
if src_dir not in sys.path:
    sys.path.insert(0, src_dir)

os.environ.setdefault("LANGSMITH_TRACING", "false")
```

### 청크 사이즈 100과 50의 차이 

```python
from pypdf import PdfReader
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter


def load_spread_pdf_as_logical_pages(pdf_path: str) -> list[Document]:
    reader = PdfReader(pdf_path)
    docs = []

    for spread_idx, page in enumerate(reader.pages):
        xs = []

        def collect_x(text, cm, tm, font_dict, font_size):
            if text.strip():
                xs.append(tm[4])

        page.extract_text(visitor_text=collect_x)

        xs = sorted(set(xs))
        gap_start, gap_end = max(
            ((xs[i], xs[i + 1]) for i in range(len(xs) - 1)),
            key=lambda pair: pair[1] - pair[0],
        )
        split_x = (gap_start + gap_end) / 2

        left_parts = []
        right_parts = []

        def collect_text(text, cm, tm, font_dict, font_size):
            stripped = text.strip()
            if not stripped:
                return

            if tm[4] < split_x:
                left_parts.append(stripped)
            else:
                right_parts.append(stripped)

        page.extract_text(visitor_text=collect_text)

        for side, parts, logical_page in (
            ("left", left_parts, spread_idx * 2 + 1),
            ("right", right_parts, spread_idx * 2 + 2),
        ):
            content = " ".join(parts).strip()
            if content:
                docs.append(
                    Document(
                        page_content=content,
                        metadata={
                            "source": pdf_path,
                            "spread_page": spread_idx,
                            "logical_page": logical_page,
                            "side": side,
                        },
                    )
                )

    return docs


docs_p4 = load_spread_pdf_as_logical_pages(file_path_p4)
print(len(docs_p4))  # 10

text_splitter_p4 = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", " •", " ▪", " ", ""],
    chunk_size=500,
    chunk_overlap=100,
    length_function=len,
)

split_docs_p4 = text_splitter_p4.split_documents(docs_p4)

for i in split_docs_p4:
    print(i.page_content)
    print("=" * 100)
```
