```python
import os
import sys
from pathlib import Path


# 프로젝트 루트 경로를 찾아 src 폴더를 sys.path에 추가합니다.
def find_project_root() -> Path:
    curr = Path.cwd()
    for parent in [curr] + list(curr.parents):
        if (parent / "pyproject.toml").exists():
            return parent
    return curr


PROJECT_ROOT = find_project_root()
src_dir = str(PROJECT_ROOT / "src")
if src_dir not in sys.path:
    sys.path.insert(0, src_dir)

os.environ.setdefault("LANGSMITH_TRACING", "false")
```

```python
from langchain_community.document_loaders import PyPDFLoader

# 프로젝트 루트를 기준으로 경로를 설정합니다.
file_path_p2 = str(PROJECT_ROOT / "data/raw/pdf/saving_tips/2.pdf")


loader_p2 = PyPDFLoader(file_path=file_path_p2)
docs_p2 = loader_p2.load()

print(f"불러온 문서 수: {len(docs_p2)}")
print(docs_p2[1].page_content)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter_p2 = RecursiveCharacterTextSplitter(
    separators=[
        "\n\n", 
        "\n", 
        " ", 
        ""], 
        chunk_size=800, chunk_overlap=200, length_function=len
)

split_docs_p2 = text_splitter_p2.split_documents(docs_p2)

for i, doc in enumerate(split_docs_p2[:3]):
    content = doc.page_content.replace("\n", " ")
    print(f"==> [Line {i:02d}]: {content}")

print(f"분할된 문서 수: {len(split_docs_p2)}")
```

```python
# 임베딩
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
embeddings


from langchain_community.vectorstores import FAISS

vectorstore_p2 = FAISS.from_documents(split_docs_p2, embeddings)
vectorstore_p2.index.ntotal

retriever_p2 = vectorstore_p2.as_retriever()
```

```python
# 질문
query_p2_food = "어르신 우대 금리 받는 법 알려줘"

# invoke() 메서드로 문서 검색
res_p2_food = retriever_p2.invoke(query_p2_food)

len(res_p2_food)
for i in res_p2_food:
    print(i.page_content.replace("\n", " "))
    print("=" * 100)
```
