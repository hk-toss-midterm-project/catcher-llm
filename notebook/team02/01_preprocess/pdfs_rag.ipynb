```python
import os
import sys
from pathlib import Path


# 프로젝트 루트 경로를 찾아 src 폴더를 sys.path에 추가합니다.
def find_project_root() -> Path:
    curr = Path.cwd()
    for parent in [curr] + list(curr.parents):
        if (parent / "pyproject.toml").exists():
            return parent
    return curr


PROJECT_ROOT = find_project_root()
src_dir = str(PROJECT_ROOT / "src")
if src_dir not in sys.path:
    sys.path.insert(0, src_dir)

os.environ.setdefault("LANGSMITH_TRACING", "false")
```

```python
from catcher_llm.retrievers.vectorstore import build_local_vectorstore, get_local_retriever

dir_path = PROJECT_ROOT / "data/raw/pdf/saving_tips"

retriever = get_local_retriever(
    chunk_size=500,
    chunk_overlap=50,
    top_k=1,
    raw_data_dir=dir_path
)

```

```python
from pathlib import Path

file_path = Path(PROJECT_ROOT / "data/raw/pdf/saving_tips/1.pdf")
dir_path = PROJECT_ROOT / "data/raw/pdf/saving_tips"

# vectorstore.py 에서 정의한 FAISS 기반 빌더 사용

vectorstore = build_local_vectorstore(
    raw_data_dir=dir_path,
    chunk_size=500,
    chunk_overlap=50
)

if vectorstore:
    if hasattr(vectorstore, "store"):
        num_docs = vectorstore.index.ntotal if hasattr(vectorstore, "index") else len(getattr(vectorstore, "store", []))
    elif hasattr(vectorstore, "index"):
        num_docs = vectorstore.index.ntotal
    else:
        num_docs = "알 수 없음"
    print(f"벡터스토어 생성 완료. 문서 수: {num_docs}")

    # 검색 기능 테스트
    query = "회사원"
    results = vectorstore.similarity_search(query, k=3)

    print(f"\n[ 검색 질의: {query} ]")
    for idx, res in enumerate(results, start=1):
        print(f"\n--- 검색 결과 {idx} ---")
        print(res.page_content)
else:
    print("생성된 벡터스토어가 없습니다.")
```
